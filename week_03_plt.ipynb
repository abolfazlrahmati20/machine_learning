```python
import matplotlib.pyplot as plt
```

```python
epoch_losses = [0.9, 0.6, 0.45, 0.38, 0.34, 0.31, 0.30]
plt.plot(epoch_losses)
plt.title("epoch losses")
plt.xlabel("epoch") 
plt.ylabel("loss")
plt.show()
```

```python
import numpy as np

labels = ['cat', 'dog', 'cat', 'cat', 'bird', 'dog']

animals, counts = np.unique(labels, return_counts=True)

plt.bar(animals, counts)
plt.title("Animal Counts")
plt.xlabel("Animal")
plt.ylabel("Count")
plt.show()
```

```python
rng = np.random.default_rng(0)
data =rng.normal(size=(2000,4))
fig, ax = plt.subplots(2, 2 )
ax=ax.flatten()
for i,a in enumerate(ax):
    a.hist(data[:,i])
    a.set_title(f"feature {i+1}")
plt.show()



```

```python

x1=np.array([5,8,4,5,8])
x2=np.array([9,3,7,11,18])
y_true = np.array([1,0,0,1,1])
y_pred = np.array([1,1,0,0,0])
c=y_true == y_pred
fig , ax = plt.subplots()
ax.scatter(x1[c],x2[c],label='correct',color='b')
ax.scatter(x1[~c],x2[~c],label='incorrect',color='r')
ax.set_title("true vs false")
ax.set_xlabel("input 1")
ax.set_ylabel("input 2")
ax.legend()
plt.show()
```
