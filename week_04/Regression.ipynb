```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
housing = fetch_california_housing(as_frame=True).frame
```

```python
housing.head()
```

```python
# x = housing[["AveRooms"]]

x = housing[["MedInc"]]
y = housing["MedHouseVal"]
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model_linear=LinearRegression()
model_linear.fit(x_train,y_train)
y_pred_linear = model_linear.predict(x_test)
```

```python
r2_score_linear = r2_score(y_test,y_pred_linear)
r2_score_linear
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
x=housing.drop("MedHouseVal",axis=1)
y = housing["MedHouseVal"]


x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
poly = PolynomialFeatures(degree=2)
x_train_poly= poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
```

```python
model_linear.fit(x_train_poly,y_train)
y_pred_linear3=model_linear.predict(x_test_poly)
r2_score_linear3= r2_score(y_test,y_pred_linear3)
r2_score_linear3
```

```python
poly2 = PolynomialFeatures(degree=5)
x_train_poly2= poly2.fit_transform(x_train)
x_test_poly2=poly2.transform(x_test)
```

```python
model_linear.fit(x_train_poly2,y_train)
y_pred_linear4 = model_linear.predict(x_test_poly2)
r2_score_linear4 = r2_score(y_test,y_pred_linear4)  
r2_score_linear4
```

```python
from sklearn.datasets import load_diabetes
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error,mean_squared_error
```

```python
diabetes= load_diabetes()
```

```python
diabetes.data
diabetes.target
```

```python
diabetes.data.shape
```

```python
df = pd.DataFrame(diabetes.data,columns=diabetes.feature_names)
df
```

```python
df["target"]=diabetes.target
```

```python
df.head()
```

```python
x=df.drop("target",axis=1)
y=df["target"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
Linear=LinearRegression()
Linear.fit(x_train,y_train)
y_pred_linear_df=Linear.predict(x_test)
```

```python
mae=mean_absolute_error(y_test,y_pred_linear_df)
mae
```

```python
rmse=np.sqrt(mean_squared_error(y_test,y_pred_linear_df))
rmse
```

```python
r2_df=r2_score(y_test,y_pred_linear_df)
r2_df
```
