```python
import numpy as np
```

```python
a = np.random.default_rng(seed=40)

b = a.integers(0, 101, (5, 5))

b[b > 50] = 0

print(b)
```

```python
a = np.random.default_rng(seed=40)

b = a.integers(0, 10, (4, 3))

print(b)

c = b.sum(axis=0)

print(c)

d = b / c

print(d)
```

```python
a = np.array([2, 5, 1])

b = np.array([4, 1, 7])

c = np.sqrt(((a - b) ** 2).sum())

d = np.linalg.norm(a - b)

print(c)

print(d)
```

```python
import time

a = np.arange(1_000_000)

start = time.time()

b = a**2 + 2*a + 1

end = time.time()

print("Vectorized Time:", end - start)

start = time.time()

c = []

for i in a:
    c.append(i**2 + 2*i + 1)

end = time.time()

print("For Loop Time:", end - start)
```
