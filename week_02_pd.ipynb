```python
import pandas as pd
```

```python
from sklearn.datasets import load_iris

bunch = load_iris(as_frame=True)

iris = bunch.frame.rename(columns={
    "sepal length (cm)": "sepal_length",
    "sepal width (cm)":  "sepal_width",
    "petal length (cm)": "petal_length",
    "petal width (cm)":  "petal_width",
})

iris["species"] = iris["target"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})

iris = iris.drop(columns="target")
iris
```

```python
iris["petal_length"].mean()
iris["petal_length"] < iris["petal_length"].mean()
iris[iris["petal_length"] < iris["petal_length"].mean()]
```

```python
a = iris[iris["petal_length"] > iris["petal_length"].mean()][["species", "petal_length"]]
print(a)
len(a)
```

```python
iris["is_big"] = (iris["petal_length"] * iris ["petal_width"]) > 10 
iris["is_big"].value_counts()
```

```python
b= iris.groupby("species").mean()

print(b)

print(b["sepal_length"].idxmax())
```

```python
from sklearn.model_selection import train_test_split

feature_cols = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

X = iris[feature_cols].to_numpy()

y = iris["species"].astype("category").cat.codes.to_numpy()

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

print(X_train.shape)
print(X_test.shape)
```
