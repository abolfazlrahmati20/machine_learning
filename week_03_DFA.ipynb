```python
housing_copy = housing.copy()
housing_copy
```

```python
import numpy as np
rng = np.random.default_rng(seed=0)
n_random =int(housing_copy.shape[0] * 0.05)
n_random
```

```python
a=rng.choice(housing_copy.index, size=n_random, replace=False)
a
```

```python
housing_copy.loc[a,"Population"]=np.nan
housing_copy["Population"].isnull().sum()
```

```python
housing_drap=housing_copy.copy()
```

```python
housing_drap.isna().sum()
```

```python
housing_drap.dropna()
```

```python
housing_drap["Population"].mean()
```

```python
housing_fill=housing_copy.copy()
housing_fill
```

```python
housing_fill["Population"].median()
```

```python
housing_fill["Population"]=housing_fill["Population"].fillna(housing_fill["Population"].median)
```
