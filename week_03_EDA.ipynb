```python
rooms = housing[["AveRooms"]]
```

```python
rooms
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
scaler=StandardScaler()
```

```python
rooms_standard= scaler.fit_transform(rooms)
rooms_standard
```

```python
MinMax_Scaler=MinMaxScaler()
```

```python
rooms_min=MinMax_Scaler.fit_transform(rooms)
rooms_min
```

```python
import matplotlib.pyplot as plt
fig , ax=plt.subplots(1,3)
ax[0].hist(rooms, bins=20)
ax[1].hist(rooms_standard, bins=20)
ax[2].hist(rooms_min, bins=20)
ax[0].set_title("room")
ax[1].set_title("Standard Scaler")      
ax[2].set_title("Min-Max Scaler")
```
